```python
social_dev_1=social_dev.copy()
social_dev_1['high_HIV']=(social_dev_1['Prevalence of HIV, female (% ages 15-24)']>0.1) \
                        & (social_dev_1['Prevalence of HIV, male (% ages 15-24)']>0.1)
px.line(social_dev_1[['high_HIV','Year','Refugee population by country or territory of asylum']]
        .groupby(['high_HIV','Year']).mean().reset_index(),
        x='Year', y='Refugee population by country or territory of asylum', color='high_HIV',
        labels={'Refugee population by country or territory of asylum': 'Refuegee Population'}, 
        title="Average refugees' population accross years")
```

```python
social_dev_2=social_dev.copy()
social_dev_2['many_refugee']=social_dev_2['Refugee population by country or territory of asylum']>=social_dev['Refugee population by country or territory of asylum'].median()
social_dev_2['avg_life_expect']=(social_dev_2['Life expectancy at birth, female (years)']+social_dev_2['Life expectancy at birth, male (years)'])/2
px.bar(social_dev_2[['many_refugee', 'avg_life_expect']].groupby('many_refugee').mean().reset_index(),
        x='many_refugee', y='avg_life_expect', color='many_refugee',
        title='Average Life Expectancy in Countries with or without many refugees')
```

```python
summary=social_dev['Refugee population by country or territory of asylum'].describe()
def map_summary(num):
    if num==np.nan:
        return num
    if num<summary['25%']:
        return 'Low'
    elif num<summary['50%']:
        return 'Normal'
    elif num<summary['75%']:
        return 'High'
    return 'Very high'
```

```python
social_dev_3=social_dev_2.copy()
social_dev_3['refugee_popu_level']=social_dev_3['Refugee population by country or territory of asylum'].apply(map_summary)
px.bar(social_dev_3[['refugee_popu_level', 'avg_life_expect']].groupby('refugee_popu_level').mean().reset_index(),
        x='refugee_popu_level', y='avg_life_expect', color='refugee_popu_level', 
        range_y=[60, 69], 
        category_orders={'refugee_popu_level': ['Low', 'Normal', 'High', 'Very high']},
        title='Average Life expectancy in countries with different refugee population level',
        color_discrete_sequence=['#c0dee5', '#61a4ad', '#07575b', '#003b46'])
```

```python
social_dev_4=social_dev_3.copy()
social_dev_4.rename(columns={'Literacy rate, youth (ages 15-24), gender parity index (GPI)': 'literacy_rate',
                             'Refugee population by country or territory of origin': 'Refugee population'}, inplace=True)
social_dev_5=social_dev_4[social_dev_4['literacy_rate']<=1].groupby('Country Code')[['literacy_rate', 'Refugee population']].mean().dropna().sort_values(by='Refugee population')
social_dev_5=social_dev_5[social_dev_5['literacy_rate']<=1]
partial=pd.concat([social_dev_5.head(3).assign(category='top 3'), social_dev_5.tail(3).assign(category='bottom 3')])
px.bar(partial.reset_index(), x='Country Code', y='literacy_rate',
        color='category', text_auto=True, title='Literacy rate in top 3 and bottom 3 countries in terms of refugee population')
```

```python
def zscore(s):
    return (s-s.mean())/s.std()
```

```python
to_stand = ['Life expectancy at birth, male (years)',
       'Life expectancy at birth, female (years)',
       'Adolescent fertility rate (births per 1,000 women ages 15-19)',
       'School enrollment, primary (gross), gender parity index (GPI)',
       'School enrollment, secondary (gross), gender parity index (GPI)',
       'School enrollment, primary and secondary (gross), gender parity index (GPI)',
       'Refugee population by country or territory of origin',
       'Labor force participation rate, total (% of total population ages 15-64) (modeled ILO estimate)',
       'Labor force participation rate, male (% of male population ages 15-64) (modeled ILO estimate)',
       'Labor force participation rate, female (% of female population ages 15-64) (modeled ILO estimate)',
       'Labor force participation rate, female (% of female population ages 15+) (modeled ILO estimate)',
       'Labor force participation rate, male (% of male population ages 15+) (modeled ILO estimate)',
       'Labor force participation rate for ages 15-24, male (%) (modeled ILO estimate)',
       'Labor force participation rate for ages 15-24, female (%) (modeled ILO estimate)',
       'Labor force participation rate for ages 15-24, total (%) (modeled ILO estimate)',
       'School enrollment, tertiary (gross), gender parity index (GPI)',
       'Unemployment, male (% of male labor force) (modeled ILO estimate)',
       'Unemployment, female (% of female labor force) (modeled ILO estimate)',
       'Vulnerable employment, male (% of male employment) (modeled ILO estimate)',
       'Vulnerable employment, female (% of female employment) (modeled ILO estimate)',
       'Proportion of seats held by women in national parliaments (%)',
       'Prevalence of HIV, female (% ages 15-24)',
       'Prevalence of HIV, male (% ages 15-24)']
final_std=social_dev[~(social_dev['Country Code'].isin(['WLD', 'IBT', 'LMY', 'IDA', 'MIC', 
                                                           'FCS', 'IDX', 'EAR', 'HPC', 'TMN', 
                                                           'ARB', 'LIC', 'MNA', 'SSA', 'PRE',
                                                           'ECS', 'HIC', 'TSS', 'AFE', 'OED',
                                                           'LDC', 'MEA', 'UMC', 'SSF', 'IBD', 
                                                           'LMC', 'SAS', 'TSA', 'LTE', 'PST', 
                                                           'TEC', 'IDB', 'ECA', 'AFW']))].copy()[to_stand]

for col in final_std.columns:
    final_std[col] = zscore(final_std[col])
final_std
```

```python
negative = ['Adolescent fertility rate (births per 1,000 women ages 15-19)',
            'Refugee population by country or territory of origin',
            'Unemployment, male (% of male labor force) (modeled ILO estimate)',
            'Unemployment, female (% of female labor force) (modeled ILO estimate)',
            'Vulnerable employment, male (% of male employment) (modeled ILO estimate)',
            'Vulnerable employment, female (% of female employment) (modeled ILO estimate)',
            'Prevalence of HIV, female (% ages 15-24)',
            'Prevalence of HIV, male (% ages 15-24)']
for col in negative:
    final_std[col] = -final_std[col]
```

```python
final['is_PAK'] = final['Country Code'] == 'PAK'
is_AFG = final[final['Country Code'] == 'PAK'].sort_values(by='Year')
percent = final[final['Country Code'] != 'PAK'].groupby('Year')[['social quality']]\
    .quantile([0.25, 0.75]).unstack().reset_index().sort_values(by='Year')
percent.columns = ['year', 'p25', 'p75']

fig = go.Figure()
fig.add_trace(go.Scatter(x=is_AFG['Year'], y=is_AFG['social quality'],
                          name='social quality of PAK',
                          line=dict(color='red')))
fig.add_trace(go.Scatter(x=percent['year'], y=percent['p75'],
                         name='75 Percentile of social quality each year',
                         line=dict(color='rgba(0,0,255,0.2)'),
                        showlegend=False,
                        hoverinfo='skip'))
fig.add_trace(go.Scatter(x=percent['year'], y=percent['p25'],
                         name='Interquartile Range of social quality',
                         fill='tonexty',
                        fillcolor='rgba(200,200,200,0.7)',
                        line=dict(color='rgba(0,0,255,0.2)'),
                        showlegend=True))
fig.update_layout(
    title='Social Quality of Afghanistan vs. the rest of the World',
    xaxis_title='Year',
    yaxis_title='Social Quality'
)
fig.show()
```

```python
def period(year):
    if year<1970:
        return '60s'
    elif year<1980:
        return '70s'
    elif year<1990:
        return '80s'
    elif year<2000:
        return '90s'
    elif year<2010:
        return '00s'
    elif year<2020:
        return '10s'
    return '20s'
```

```python
final['period'] = final['Year'].apply(period)
```

```python
final['log quality'] = final['social quality'].apply(lambda x: abs(x)).apply(np.log)
```
